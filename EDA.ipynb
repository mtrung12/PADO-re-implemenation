```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from transformers import AutoTokenizer
from tqdm import tqdm

# 1. Load your dataset (replace with your actual file path)
df = pd.read_csv('data/processed_essay.csv')
# For demonstration, assuming your dataframe is called 'df' and text column is 'text'

# 2. Basic Character and Word Counts
print("Calculating word and character counts...")
df['char_count'] = df['text'].apply(lambda x: len(str(x)))
df['word_count'] = df['text'].apply(lambda x: len(str(x).split()))




print("\n--- Percentiles (The OOM Danger Zone) ---")
percentiles = [0.50, 0.75, 0.90, 0.95, 0.99, 0.999]
for p in percentiles:
    char_percentile = df['char_count'].quantile(p)
    word_percentile = df['word_count'].quantile(p)
    print(f"{int(p*100)}th Percentile - Characters: {char_percentile}, Words: {word_percentile}")

# 5. Visualizing the Distribution
plt.figure(figsize=(15, 5))

# Plot Word Count
plt.subplot(1, 2, 1)
sns.histplot(df['word_count'], bins=50, kde=True, color='skyblue')
plt.title('Word Count Distribution')
plt.xlabel('Number of Words')
plt.ylabel('Frequency')

```
